```python
import numpy as np

num_frames = 200

# Random initial orientation
euler0 = np.random.uniform(0, 2*np.pi, 3)

# Constant angular velocity (rad/frame)
omega = np.array([
    np.deg2rad(4.0),   # X: 1° per frame
    np.deg2rad(0.7),   # Y: 0.5° per frame
    np.deg2rad(6.0)    # Z: 2° per frame
])

with open("rotations.txt", "w") as f:
    for i in range(num_frames):
        euler = (euler0 + i * omega) % (2*np.pi)
        f.write(f"{euler[0]:.6f} {euler[1]:.6f} {euler[2]:.6f}\n")
```

```python
import numpy as np
from scipy.spatial.transform import Rotation as R

num_frames = 100

# Random initial translation
t0 = np.random.uniform(-0.2, 0.2, 3)

# Constant translation per frame (meters/frame)
v = np.array([
    0.002,
    -0.001,
    0.0005
])

# Random initial orientation
q = R.random()

# Constant angular velocity (rad/frame)
omega = np.array([
    np.deg2rad(2.0),
    np.deg2rad(0.7),
    np.deg2rad(1.0)
])

dq = R.from_rotvec(omega)

with open("poses.txt", "w") as f:
    t = t0.copy()

    for _ in range(num_frames):

        quat = q.as_quat()      # (x, y, z, w)

        f.write(
            f"{t[0]:.6f} {t[1]:.6f} {t[2]:.6f} "
            f"{quat[0]:.6f} {quat[1]:.6f} "
            f"{quat[2]:.6f} {quat[3]:.6f}\n"
        )

        t += v
        q = q * dq
```
